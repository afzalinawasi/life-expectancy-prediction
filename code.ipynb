Therefore, transformation was checked and considered, but not applied blindly. This keeps the features interpretable while allowing model performance and cross-validation results to guide the final decision.

### Outlier check using IQR method

```python
numeric_features = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Remove target because we already checked target outliers separately
if target in numeric_features:
    numeric_features.remove(target)

outlier_summary = []

for col in numeric_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    outlier_percentage = (outlier_count / len(df)) * 100
    
    outlier_summary.append({
        "column": col,
        "outlier_count": outlier_count,
        "outlier_percentage": outlier_percentage
    })

outlier_summary_df = pd.DataFrame(outlier_summary)
outlier_summary_df = outlier_summary_df.sort_values(by="outlier_percentage", ascending=False)

display(outlier_summary_df)
```

```python
# Boxplots for top 6 columns with highest outlier percentage

top_outlier_cols = outlier_summary_df.head(6)["column"].tolist()

for col in top_outlier_cols:
    plt.figure(figsize=(8, 2.5))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.show()
```

#### Outlier analysis summary

The dataset has visible outliers in several numerical features. The highest outlier percentages are seen in `Measles` and `HIV/AIDS` at around 18.45%, followed by `under-five deaths`, `percentage expenditure`, `GDP`, and `infant deaths`.
The boxplots confirm strong right-side outliers, especially for disease counts, mortality counts, health expenditure, and GDP. These values should not be removed blindly because they may represent valid real-world country-year records.

#### Outlier treatment decision

The IQR method and boxplots show visible outliers in several numerical features, especially Measles, HIV/AIDS, under-five deaths, percentage expenditure, GDP, infant deaths, and Population.
These outliers are not removed automatically because they may represent valid real-world country-year records. For example, countries with very large populations, disease outbreaks, or high health expenditure can naturally produce extreme values.

**Instead of deleting or capping these records, the following treatment decision is applied:**

- No rows are removed only because of outlier values.
- No separate outlier capping is applied at this stage.
- RobustScaler is used during preprocessing because it scales features using the median and interquartile range, making it less sensitive to extreme values than StandardScaler or MinMaxScaler.
- Tree-based models such as Decision Tree, Random Forest, and XGBoost are included because they can usually handle outliers better than purely distance-based or linear models.
- Model performance, overfitting checks, and cross-validation are used to confirm whether the retained outliers are affecting generalization.

Therefore, outliers were checked and treated through robust preprocessing rather than removed blindly.

### Next step and approach

For model preparation, rows with missing `Life expectancy` will be removed because the target value is required for supervised regression. The `Country` column will be excluded in the first model version to avoid high-cardinality encoding and reduce the risk of the model memorizing country names instead of learning health and economic patterns.

The dataset will be split into train and test sets before imputation or scaling to avoid data leakage. Missing numerical values will be handled using median imputation because several features are skewed and contain outliers. Scaling will be applied for scale-sensitive models such as SVM and KNN, while tree-based models such as Decision Tree, Random Forest, and XGBoost will be trained and compared separately.

```python
from sklearn.model_selection import train_test_split

#1. Create modelling dataset
df_model = df.copy()

#2. Remove rows where target is missing
df_model = df_model.dropna(subset=[target])

#3. Drop Country for first model version to avoid high-cardinality encoding
df_model = df_model.drop(columns=["Country"])

# Define X and y
X = df_model.drop(columns=[target])
y = df_model[target]

#4. Train-test split: 80% train, 20% test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nColumns used for modelling:")
print(X.columns.tolist())

print("\nMissing values in X_train:")
display(X_train.isnull().sum().sort_values(ascending=False))

print("\nTarget summary in y_train:")
display(y_train.describe())
```

### Next step: preprocessing pipeline

```python
from sklearn.preprocessing import RobustScaler

X_train_processed = X_train.copy()
X_test_processed = X_test.copy()

status_mapping = {
    "Developing": 0,
    "Developed": 1
}

X_train_processed["Status"] = X_train_processed["Status"].map(status_mapping)
X_test_processed["Status"] = X_test_processed["Status"].map(status_mapping)

#1. Identify numerical columns
numeric_cols = X_train_processed.select_dtypes(include=["int64", "float64"]).columns.tolist()

#2. Fill missing numerical values using training median only
train_medians = X_train_processed[numeric_cols].median()

X_train_processed[numeric_cols] = X_train_processed[numeric_cols].fillna(train_medians)
X_test_processed[numeric_cols] = X_test_processed[numeric_cols].fillna(train_medians)

#3. Scale numerical columns
scaler = RobustScaler()

X_train_processed[numeric_cols] = scaler.fit_transform(X_train_processed[numeric_cols])
X_test_processed[numeric_cols] = scaler.transform(X_test_processed[numeric_cols])

print("Manual preprocessing completed.")

print("\nMissing values in X_train_processed:")
display(X_train_processed.isnull().sum().sort_values(ascending=False))

print("\nMissing values in X_test_processed:")
display(X_test_processed.isnull().sum().sort_values(ascending=False))

print("\nProcessed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)

display(X_train_processed.head())
```

#### Preprocessing insight summary

- Missing values are now 0 in both training and testing datasets.
- Shape is preserved: 2342 × 20 for training and 586 × 20 for testing.
- Status has been encoded as Developed = 1, Developing = 0.
- Numerical features were imputed using training medians only, then scaled using RobustScaler, which is suitable because the dataset has strong outliers.

### Model training and evaluation

```python
print("X_train_processed exists:", "X_train_processed" in globals())
print("X_test_processed exists:", "X_test_processed" in globals())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Define regression models
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42, n_estimators=100),
    "SVM": SVR(),
    "KNN": KNeighborsRegressor(),
    "XGBoost": XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_estimators=100
    )
}

# Train and evaluate models
results = []

for model_name, model in models.items():
    model.fit(X_train_processed, y_train)
    
    train_pred = model.predict(X_train_processed)
    test_pred = model.predict(X_test_processed)
    
    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
    
    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)
    
    results.append({
        "Model": model_name,
        "Train MAE": mean_absolute_error(y_train, train_pred),
        "Test MAE": mean_absolute_error(y_test, test_pred),
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "Train R2": train_r2,
        "Test R2": test_r2,
        "Overfitting Gap": train_r2 - test_r2
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="Test R2", ascending=False)

display(results_df)
```

#### Model training and evaluation summary

Six supervised regression models were trained to predict life expectancy: Linear Regression, Decision Tree, Random Forest, SVM, KNN, and XGBoost. The models were compared using MAE, RMSE, R² score, and the train-test R² gap to check both prediction accuracy and overfitting.

- Random Forest achieved the best overall performance with the highest Test R² of 0.966979, the lowest Test RMSE of 1.690046, and the lowest Test MAE of 1.055229. This means Random Forest explains around 96.7% of the variation in life expectancy on unseen test data, with an average prediction error of close to 1 year.

- XGBoost performed almost equally well with a Test R² of 0.965664 and Test RMSE of 1.723380. However, Random Forest slightly outperformed XGBoost on both MAE and RMSE, making it the stronger model in this initial comparison.

- Decision Tree achieved a perfect Train R² of 1.0 but dropped to a Test R² of 0.920800. This indicates overfitting, as the model learned the training data too closely but generalized less effectively on unseen data.

- KNN, Linear Regression, and SVM performed weaker than the ensemble models. This suggests that life expectancy prediction depends on non-linear relationships between health, economic, demographic, and education-related indicators, which are better captured by ensemble tree-based models.

Based on the initial train-test evaluation, **R**andom Forest is the best-performing model so far**. However, cross-validation should be performed before final model selection to confirm that its performance is stable across different data splits.

### Outlier and transformation treatment decision

- Outliers were checked using the IQR method and boxplots. Several features such as `Measles`, `HIV/AIDS`, `GDP`, `Population`, `infant deaths`, and `under-five deaths` showed strong outliers.

- These outliers were not removed blindly because they may represent valid real-world country-year observations, especially for countries with large populations, high disease burden, or unusual health conditions. Removing them could reduce the real-world meaning of the dataset.

- For skewed and outlier-heavy numerical features, RobustScaler was used during preprocessing. RobustScaler uses the median and interquartile range, so it reduces the impact of extreme values better than MinMaxScaler or StandardScaler.

- No separate log transformation or outlier capping was applied at this stage because tree-based ensemble models such as Random Forest and XGBoost can handle non-linear patterns and outliers better than linear models. The final decision is validated through model performance, overfitting checks, and cross-validation.

### Overfitting treatment using tuned models

```python
tuned_models = {
    "Tuned Decision Tree": DecisionTreeRegressor(
        random_state=42,
        max_depth=6,
        min_samples_split=10,
        min_samples_leaf=5
    ),

    "Tuned Random Forest": RandomForestRegressor(
        random_state=42,
        n_estimators=100,
        max_depth=12,
        min_samples_split=5,
        min_samples_leaf=2
    ),

    "Tuned XGBoost": XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_estimators=300,
        learning_rate=0.05,
        max_depth=3,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0
    )
}

tuned_results = []

for model_name, model in tuned_models.items():
    model.fit(X_train_processed, y_train)

    train_pred = model.predict(X_train_processed)
    test_pred = model.predict(X_test_processed)

    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))

    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)

    tuned_results.append({
        "Model": model_name,
        "Train MAE": mean_absolute_error(y_train, train_pred),
        "Test MAE": mean_absolute_error(y_test, test_pred),
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "Train R2": train_r2,
        "Test R2": test_r2,
        "Overfitting Gap": train_r2 - test_r2
    })

tuned_results_df = pd.DataFrame(tuned_results)
tuned_results_df = tuned_results_df.sort_values(by="Test R2", ascending=False)

display(tuned_results_df)
```

#### Observation and next step: 

Tuning helped reduce overfitting, especially for Decision Tree and XGBoost. However, the original Random Forest still performs slightly better than the tuned Random Forest on test accuracy.

So we should not automatically choose the tuned model. Next, we should run cross-validation on the strongest candidates:
- Original Random Forest
- Tuned Random Forest
- Original XGBoost
- Tuned XGBoost

This will tell us which model is more stable across different splits, not just one train-test split.

### Cross-validation for top models

```python
from sklearn.model_selection import KFold
from sklearn.base import clone

# Manual preprocessing function for each cross-validation fold
def preprocess_fold(X_train_fold, X_valid_fold):
    X_train_fold_processed = X_train_fold.copy()
    X_valid_fold_processed = X_valid_fold.copy()
    
    # Encode Status
    status_mapping = {
        "Developing": 0,
        "Developed": 1
    }
    
    X_train_fold_processed["Status"] = X_train_fold_processed["Status"].map(status_mapping)
    X_valid_fold_processed["Status"] = X_valid_fold_processed["Status"].map(status_mapping)
    
    # Identify numerical columns except Status
    numeric_cols_fold = X_train_fold_processed.select_dtypes(include=["int64", "float64"]).columns.tolist()
    
    if "Status" in numeric_cols_fold:
        numeric_cols_fold.remove("Status")
    
    # Fill missing values using fold training median only
    fold_medians = X_train_fold_processed[numeric_cols_fold].median()
    
    X_train_fold_processed[numeric_cols_fold] = X_train_fold_processed[numeric_cols_fold].fillna(fold_medians)
    X_valid_fold_processed[numeric_cols_fold] = X_valid_fold_processed[numeric_cols_fold].fillna(fold_medians)
    
    # Scale using fold training data only
    fold_scaler = RobustScaler()
    
    X_train_fold_processed[numeric_cols_fold] = fold_scaler.fit_transform(X_train_fold_processed[numeric_cols_fold])
    X_valid_fold_processed[numeric_cols_fold] = fold_scaler.transform(X_valid_fold_processed[numeric_cols_fold])
    
    return X_train_fold_processed, X_valid_fold_processed


# Models selected for cross-validation
cv_models = {
    "Original Random Forest": RandomForestRegressor(
        random_state=42,
        n_estimators=100
    ),

    "Tuned Random Forest": RandomForestRegressor(
        random_state=42,
        n_estimators=10,
        max_depth=12,
        min_samples_split=5,
        min_samples_leaf=2
    ),

    "Original XGBoost": XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_estimators=100
    ),

    "Tuned XGBoost": XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_estimators=30,
        learning_rate=0.05,
        max_depth=3,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0
    )
}

# 5-fold cross-validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

cv_results = []

for model_name, model in cv_models.items():
    fold_mae_scores = []
    fold_rmse_scores = []
    fold_r2_scores = []
    
    for train_index, valid_index in kf.split(X_train):
        X_train_fold = X_train.iloc[train_index]
        X_valid_fold = X_train.iloc[valid_index]
        y_train_fold = y_train.iloc[train_index]
        y_valid_fold = y_train.iloc[valid_index]
        
        X_train_fold_processed, X_valid_fold_processed = preprocess_fold(
            X_train_fold,
            X_valid_fold
        )
        
        fold_model = clone(model)
        fold_model.fit(X_train_fold_processed, y_train_fold)
        
        valid_pred = fold_model.predict(X_valid_fold_processed)
        
        fold_mae_scores.append(mean_absolute_error(y_valid_fold, valid_pred))
        fold_rmse_scores.append(np.sqrt(mean_squared_error(y_valid_fold, valid_pred)))
        fold_r2_scores.append(r2_score(y_valid_fold, valid_pred))
    
    cv_results.append({
        "Model": model_name,
        "CV MAE Mean": np.mean(fold_mae_scores),
        "CV RMSE Mean": np.mean(fold_rmse_scores),
        "CV R2 Mean": np.mean(fold_r2_scores),
        "CV R2 Std": np.std(fold_r2_scores)
    })

cv_results_df = pd.DataFrame(cv_results)
cv_results_df = cv_results_df.sort_values(by="CV R2 Mean", ascending=False)

display(cv_results_df)
```

#### Cross-validation summary

Cross-validation was performed on the strongest candidate models: Original Random Forest, Tuned Random Forest, Original XGBoost, and Tuned XGBoost.

The Original Random Forest was evaluated using the same configuration as the final selected model, with `n_estimators=100` and `random_state=42`.

Original Random Forest achieved the best cross-validation performance with the highest CV R² mean of 0.9567, the lowest CV RMSE mean of 1.9889, and the lowest CV MAE mean of 1.2570. This confirms that the model performs consistently across different training-validation splits.

Original XGBoost was also stable and performed very close to Random Forest, but its CV MAE and CV RMSE were slightly higher. Tuned Random Forest did not improve the overall score, and Tuned XGBoost showed reduced performance.

Based on both train-test evaluation and cross-validation, **Original Random Forest is the strongest final model candidate.**
